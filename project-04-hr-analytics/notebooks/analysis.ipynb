```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/WA_Fn-UseC_-HR-Employee-Attrition.csv')
print(df.shape)
print(df['Attrition'].value_counts())
```

```python
attrition_counts = df['Attrition'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(attrition_counts.values, 
        labels=attrition_counts.index, 
        autopct='%1.1f%%',
        colors=['steelblue', 'tomato'])
plt.title('Employee Attrition Rate', fontsize=14)
plt.tight_layout()
plt.savefig('../visuals/attrition_rate.png')
plt.show()
```

## What it shows: The percentage of employees who left vs. those who stayed.


```python
dept_attrition = df.groupby(['Department', 'Attrition']).size().unstack()

dept_attrition.plot(kind='bar', figsize=(9,5), 
                    color=['steelblue','tomato'])
plt.title('Attrition by Department', fontsize=14)
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.legend(['No', 'Yes'])
plt.tight_layout()
plt.savefig('../visuals/attrition_by_department.png')
plt.show()
```

## What it shows: A comparison of stays and departures across different departments.
## Why it matters: Pinpoints which specific business units are "hotspots" for turnover and need cultural or structural interventions.

```python
overtime_attrition = df.groupby(['OverTime', 'Attrition']).size().unstack()

overtime_attrition.plot(kind='bar', figsize=(7,5),
                        color=['steelblue','tomato'])
plt.title('Overtime vs Attrition', fontsize=14)
plt.xlabel('Overtime')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.legend(['No', 'Yes'])
plt.tight_layout()
plt.savefig('../visuals/overtime_vs_attrition.png')
plt.show()
```

## What it shows: The correlation between working extra hours and the decision to leave the company.
## Why it matters: Directly measures if heavy workloads and potential burnout are driving talent away.
